Bring all six feature comparisons together here. For each feature:

### Age groups

Compare `age_group` with `Survived` using your Data-step choices.


```python
# Create and display the count cross-tab.
pd.crosstab(titanic_df["age_group"],titanic_df["Survived"])
```

```python
# Calculate conditional percentages and display only the survived column. - within comparison group - index (CONDITIONAL)
age_group_survival_percent = pd.crosstab(titanic_df["age_group"], titanic_df["Survived"], normalize="index").mul(100).round(1)
age_group_survival_percent[[1]]
```

For age groups, also display joint percentages (`normalize="all"`, multiplied by 100) and make a count chart showing both outcomes. Compare this with the survivor-percentage chart to distinguish counts, joint percentages, and within-group percentages.


```python
# Display joint percentages, then create a count chart in a separate cell.
joint_age_group_survival_percent = pd.crosstab(titanic_df["age_group"], titanic_df["Survived"], normalize="all").mul(100).round(1)
joint_age_group_survival_percent#[[1]]
```

```python
# Plot the conditional percentage who survived.
age_group_survival_percent[1].plot(kind="bar", color="darkseagreen", legend=True)
plt.title("Titanic Survival Percentage by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Percent who survived within each age group")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(labels=['Survived'], loc="upper left", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
